# PoisonShieldAI — CPU/GPU quick start

This notebook trains a complete **offline, synthetic-data** attacker/defender/victim workflow. It is a software smoke test, not an empirical claim about CIC-IDS2017 or TON_IoT.


```python
from pathlib import Path
import sys
repo = Path.cwd()
if not (repo/'src'/'poisonshield').exists():
    raise RuntimeError('Run from the extracted PoisonShieldAI repository root (or set repo manually).')
sys.path.insert(0, str(repo/'src'))
```

```python
from poisonshield.pipeline import read_config, ExperimentSession
cfg = read_config(repo/'configs'/'quick_experiment.yaml')
cfg['experiment']['output_dir'] = str(repo/'outputs'/'notebook')
cfg['experiment']['synthetic_rows'] = 600
cfg['experiment']['victim'] = 'all'
runner = ExperimentSession(cfg)
print('Feature count:', runner.data.schema.n_features)
print('Classes:', runner.data.classes)
```

```python
results = runner.run(seed=0, rate=0.10, victim='all')
display(results[['victim', 'method', 'status', 'accuracy', 'macro_f1', 'clean_retention']])
```

```python
from IPython.display import Image, display
image_file = next((repo/'outputs'/'notebook').rglob('rf_method_macro_f1.png'))
display(Image(filename=str(image_file), width=900))
```
